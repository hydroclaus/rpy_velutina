```python
import asyncio
from concurrent.futures import ThreadPoolExecutor
import time
```

```python
def calculate_test(x, y):
    return x + y
```

```python
x = [1, 2, 3, 4, 5]
y = 1
z = [5, 3, 6, 1, 2]
```

```python
with ThreadPoolExecutor(max_workers=1) as executor:
    results = list(executor.map(lambda x: calculate_test(x, y), x))
```

```python
results
```

```python
async def calculate_test(x, y, z):
    await asyncio.sleep(z)
    return x + y

async def main():
    tasks = [calculate_test(xi, y, zi) for xi, zi in zip(x, z)]
    results = await asyncio.gather(*tasks)
    print(results)
```

```python
await main()
```
