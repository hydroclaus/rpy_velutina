```python
import asyncio
from concurrent.futures import ThreadPoolExecutor
import time
```

```python
def calculate_test(x, y):
    return x + y
```

```python
x = [1, 2, 3, 4, 5]
y = 1
z = [5, 3, 6, 1, 2]
```

```python
with ThreadPoolExecutor(max_workers=1) as executor:
    results = list(executor.map(lambda x: calculate_test(x, y), x))
```

```python
results
```

```python
async def calculate_test(x, y, z):
    await asyncio.sleep(z)
    return x + y

async def main():
    tasks = [calculate_test(xi, y, zi) for xi, zi in zip(x, z)]
    results = await asyncio.gather(*tasks)
    print(results)
```

```python
await main()
```

```python
# Sättigungsdampfdruck p_s(T)
T = 25 # °C
T_K = T + 273.15 # K
p_s = 6.1078 * math.exp( (17.27 * T) / (T + 237.3) ) * 100 # Pa

# Wasserdampfdruck p_w
relHum = 60 # %
phi = relHum / 100
p_w = phi * p_s # Pa

# Absolute Luftfeuchtigkeit rho_w
R_w = 461.52 # J/kg*K (Spezifische Gaskonstante von Wasserdampf)
rho_w = (p_w / (R_w * T_K)) * 1000 # g/m^3

print(p_w)
print(rho_w)
```

```python
def calculate_abs_H(temperature, relHum):
    T_K = temperature + 273.15 # K
    p_s = 6.1078 * math.exp( (17.27 * temperature) / (temperature + 237.3) ) * 100 # Sättigungsdampfdruck in Pa
    phi = relHum / 100
    p_w = phi * p_s # Wasserdampfdruck in Pa
    R_w = 461.52 # Spezifische Gaskonstante von Wasserdampf in J/kg*K
    rho_w = (p_w / (R_w * T_K)) * 1000 # Absolute Luftfeuchtigkeit in g/m^3
    return round(rho_w, 3)
```

```python
calculate_abs_H(35, 100)
```
